```python
import pandas as pd
import numpy as np
import requests
import io
```

```python
# original data used for xmatch
data_from_drive = pd.read_csv('full_galactic_table.csv')
```

```python
data_from_drive.shape
```

```python
# results from the xmatch,unique objects
data = pd.read_parquet('../data/redo_data_Sep2026/full_galactic_table_matches.parquet')
```

```python
data.shape
```

```python
#percentage of xmatch returned
np.round(100*data.shape[0]/data_from_drive.shape[0], 3)
```

```python
# add column with folder name
name_list = []

for i in range(data.shape[0]):
    name = data.iloc[i]['full_id'].split('/', 1)[0]

    if 'LEGUSv3' in name:
        name_use = 'LEGUSv3'
    elif name == 'nan':
        name_use = data.iloc[i]['full_id']
    else:
        name_use = name

    name_list.append(name_use)

data['folder'] = name_list
```

```python
# calculate number of xmatches per folder
folder, pop =  np.unique(data['folder'].values, return_counts=True)

print('Folder \t n_matches')
for i in range(len(folder)):
    print(folder[i], '\t', pop[i])
```

```python
# nan are a result of a problem in the database, these were checked by hand -- nothing happening
for name in data[data['full_id'].apply(lambda x: 'nan' in x)]['objectId'].values:
    print('https://ztf.fink-portal.org/' + name)
```

```python
# focus on the galactic folder
data_galactic = data[data['folder'].apply(lambda x: 'galactic' in x)]
```

```python
class_gal, popo_gal = np.unique(data_galactic['finkclass'].values, return_counts=True)

for i in range(class_gal.shape[0]):
    print(class_gal[i], '\t', popo_gal[i])
```

```python
# get data for Unknown -- API should not be used by large numbers, so only first 500
mask = data_galactic['finkclass'] == 'Unknown'
obj_list = list(data_galactic[mask.values]['objectId'].values)


r = requests.post(
  "https://api.ztf.fink-portal.org/api/v1/objects",
  json={
    "objectId":  ','.join(obj_list[:500]),
    "columns": "i:objectId,i:jd,i:magpsf,i:sigmapsf,d:cdsxmatch",
    "output-format": "json" 
  }
)

pdf = pd.read_json(io.BytesIO(r.content))
```

```python
# almost everything is spurious
names, counts = np.unique(pdf['i:objectId'].values, return_counts=True)
indx = np.argsort(counts)

for i in range(indx.shape[0]):
    if counts[indx[i]] > 3:
        print('https://ztf.fink-portal.org/' + names[indx[i]], counts[indx[i]])
```
