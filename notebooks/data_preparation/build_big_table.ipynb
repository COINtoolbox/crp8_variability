# YMCs folder

In this folder ra, dec columns were infered, because there were no headers in the files

```python
flist_ymcs = os.listdir('../data/redo_data_Sep2026/ymcs/')

data_ymcs_list = []

for fname in progressbar(flist_ymcs):

    if '.csv' in fname:
        data_ymcs_temp = pd.read_csv('../data/redo_data_Sep2026/ymcs/' + fname)
        if data_ymcs_temp.shape[0] > 0:
            if min(data_ymcs_temp['4']) >= 0 and max(data_ymcs_temp['4']) < 360 and \
               min(data_ymcs_temp['5']) >=-90 and max(data_ymcs_temp['5']) < 90:
                mask = data_ymcs_temp['5'].values > -28
                if sum(mask) > 0:
                    data_ymcs_temp['source'] = 'ymcs'
                    data_ymcs_temp.rename(columns={'4':'ra', '5':'dec'}, inplace=True)
                    data_ymcs_list.append(data_ymcs_temp[mask])
            elif min(data_ymcs_temp['2']) >= 0 and max(data_ymcs_temp['2']) < 360 and \
                 min(data_ymcs_temp['3']) >=-90 and max(data_ymcs_temp['3']) < 90:
                mask = data_ymcs_temp['3'].values > -28
                if sum(mask) > 0:
                    data_ymcs_temp.rename(columns={'2':'ra', '3':'dec'}, inplace=True)
                    data_ymcs_list.append(data_ymcs_temp[mask])

            elif min(data_ymcs_temp['3']) >= 0 and max(data_ymcs_temp['3']) < 360 and \
                 min(data_ymcs_temp['4']) >=-90 and max(data_ymcs_temp['4']) < 90:
                mask = data_ymcs_temp['4'].values > -28
                if sum(mask) > 0:
                    data_ymcs_temp.rename(columns={'3':'ra', '4':'dec'}, inplace=True)
                    data_ymcs_list.append(data_ymcs_temp[mask])
            elif min(data_ymcs_temp['6']) >= 0 and max(data_ymcs_temp['6']) < 360 and \
                 min(data_ymcs_temp['7']) >=-90 and max(data_ymcs_temp['7']) < 90:
                mask = data_ymcs_temp['7'].values > -28
                if sum(mask) > 0:
                    data_ymcs_temp.rename(columns={'6':'ra', '7':'dec'}, inplace=True)
                    data_ymcs_list.append(data_ymcs_temp[mask])
            else:
                print(fname)

data_ymcs = pd.concat(data_ymcs_list, ignore_index=True)
```

```python
data_ymcs.shape
```

```python
data_ymcs['madeup_id'] = np.arange(data_ymcs.shape[0])
```

```python
data_ymcs['full_id'] = [str(data_ymcs.iloc[i]['source']) + '/' + str(data_ymcs.iloc[i]['madeup_id']) 
                               for i in range(data_ymcs.shape[0])]
```

```python
data_ymcs_join_ini = deepcopy(data_ymcs[['full_id', 'ra', 'dec']])
```

```python
data_ymcs_join_ini.drop_duplicates(subset=['ra', 'dec']).shape[0] - data_ymcs_join_ini.shape[0]
```

```python
data_ymcs_join = data_ymcs_join_ini.drop_duplicates(subset=['ra', 'dec'], keep='first')
```

```python
data_ymcs_join.shape
```

```python
data_ymcs_join.drop_duplicates(subset=['ra', 'dec']).shape[0] - data_ymcs_join.shape[0]
```

```python
data_ymcs_join.keys()
```

```python
min(data_ymcs_join['ra']) >= 0 and max(data_ymcs_join['ra']) < 360 
```

## redo_csv folder

## GalGCs_GAIAFlags folder

```python
flist_gaia = os.listdir('../data/redo_data_Sep2026/GalGCs_GAIAFlags/')

data_gaia_list = []

columns_gaia = ['Unnamed: 0',
 'source_id',
 'ra',
 'dec',
 'x',
 'y',
 'plx',
 'pmra',
 'pmdec',
 'plxe',
 'pmrae',
 'pmdece',
 'pmcorr',
 'g_mag',
 'bp_rp',
 'Sigma',
 'qflag',
 'memberprob',
 'source']

for fname in progressbar(flist_gaia):
    if '.csv' in fname:

        data_gaia_temp = pd.read_csv('../data/redo_data_Sep2026/GalGCs_GAIAFlags/' + fname, index_col=False)
    
        if data_gaia_temp.shape[0] > 0:

            if '#' in data_gaia_temp.keys():
                data_gaia_temp = pd.read_csv('../data/redo_data_Sep2026/GalGCs_GAIAFlags/' + fname, index_col=False, skiprows=1,
                                        names=columns_gaia)

            if 'RA_ICRS' in data_gaia_temp.keys():
                data_gaia_temp.rename(columns={'RA_ICRS':'ra', 'DE_ICRS':'dec'}, inplace=True)

            data_gaia_temp.drop_duplicates(subset=['ra', 'dec'], keep='first', inplace=True)
            mask_dec = data_gaia_temp['dec'] > -28
            if sum(mask_dec) > 0:

                if 'pCat' in data_gaia_temp.keys():
                    data_gaia_temp['name'] = data_gaia_temp['GAIA_ID'] 
                elif 'Source' in data_gaia_temp.keys():
                    data_gaia_temp['name'] = data_gaia_temp['Source']
                else:
                    data_gaia_temp['name'] = data_gaia_temp['source_id']
            
                data_gaia_temp['source'] = 'GalGCs_GAIAFlags/' + fname

                # rename ra,dec columns
                data_gaia_list.append(data_gaia_temp[mask_dec])

data_gaia = pd.concat(data_gaia_list, ignore_index=True)
```

```python
data_gaia.shape
```

```python
# check for duplicates
data_gaia.drop_duplicates(subset=['ra', 'dec']).shape[0] - data_gaia.shape[0]
```

```python
data_gaia.drop_duplicates(subset=['ra', 'dec'], keep='first', inplace=True)
```

```python
data_gaia.drop_duplicates(subset=['ra','dec']).shape[0] == data_gaia.shape[0]
```

```python
data_gaia['full_id'] = [str(data_gaia.iloc[i]['source']) + '/' + str(data_gaia.iloc[i]['name']) 
                               for i in range(data_gaia.shape[0])]
```

```python
data_gaia_join = data_gaia[['full_id', 'ra', 'dec']]
```
